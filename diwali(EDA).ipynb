```python
sf=df.groupby(['Gender'],as_index= False)['Amount'].sum().sort_values(by='Amount',ascending=False)
sns.barplot(x='Gender',y='Amount',data=sf)
```

```python
sns.countplot(x='Age Group',data=df,hue='Gender')
```

```python
ax=sns.countplot(x='Age Group',data=df,hue='Gender')

for bars in ax.containers:
    ax.bar_label(bars)
```

```python
sales_age= df.groupby(['Age Group'],as_index=False)['Amount'].sum().sort_values(by='Amount',ascending=False)
sns.barplot(x='Age Group',y='Amount',data = sales_age,hue='Age Group')
```

```python
#top no.of order by top 10 state
sales_age= df.groupby(['State'],as_index=False)['Orders'].sum().sort_values(by='Orders',ascending=False).head(15)
sns.set(rc=({'figure.figsize':(15,5)}))
sns.barplot(x='State',y='Orders',data = sales_age,hue='State')
```

```python
sales_age= df.groupby(['State'],as_index=False)['Amount'].sum().sort_values(by='Amount',ascending=False).head(15)
sns.set(rc=({'figure.figsize':(15,6)}))
sns.barplot(x='State',y='Amount',data = sales_age,hue='Amount')
```

```python
sales_state= df.groupby(['Marital_Status','Gender'],as_index=False)['Amount'].sum().sort_values(by='Amount',ascending=False).head(15)
sns.set(rc=({'figure.figsize':(15,5)}))
sns.barplot(x='Marital_Status',y='Amount',data = sales_state,hue='Gender')
```
